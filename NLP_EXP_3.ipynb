```python
print("Mulla Huzaifa", "UIN:231A017\n")
import nltk
nltk.download('punkt_tab')
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
from nltk import word_tokenize
tokenized=word_tokenize(text_emoji)
print("Tokenized words:", tokenized)
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
import string
punctuations = list(string.punctuation)
print(punctuations)
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
print("With punctuation:",tokenized)
print()
text_punct =[]
for word in tokenized:
    if word not in punctuations:
        text_punct.append(word)
print("Without punctuation:",text_punct)
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
print((stopwords.words('english')))
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
text_clean=[]
for word in text_punct:
    if word not in stopwords.words('english'):
        text_clean.append(word)
print(text_clean)
```

```python
print("Mulla Huzaifa", "UIN:231A017\n")
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
print( [stemmer.stem(word) for word in text_clean])
```
