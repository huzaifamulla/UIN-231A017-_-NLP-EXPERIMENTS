```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
import nltk
from nltk import word_tokenize, bigrams, FreqDist, ConditionalFreqDist
from nltk.corpus import stopwords
from collections import defaultdict

#Download necessary NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

#Step 1: Load and preprocess the text corpus
def preprocess(text):
    #convert to lower case
    text = text.lower()
    #tokenization
    tokens = word_tokenize(text)
    #remove punctuations and stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token.isalnum() and token not in stop_words]
    return tokens
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#The quick brown fox jumps over the lazy dog. The fox is quick and dog is lazy.
corpus = "Science is a most fast growing field. Engineering and architecture make the basic necessity of survival. Engineering is one of the most opted technology field. Engineering has contributed in a large manner. Architecture provided the basic foundation of human civilization.Engineering and architecture are hence required to sustain humanity"
test = "Engineering is a fast grwoing field"
tokens = preprocess(corpus)
print(tokens)
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Generate Bigrams
bigrams_list = list(bigrams(tokens))
print(bigrams_list)
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Calculate bigram frequency
bigrams_freq = FreqDist(bigrams_list)
bigrams_freq
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Calculate conditional frequencies to predict next word
cfd = ConditionalFreqDist(bigrams_list)
print(cfd.items())
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
from nltk import trigrams
trigrams_list = list(trigrams(tokens))
print(trigrams_list)
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
trigrams_freq = FreqDist(trigrams_list)
trigrams_freq
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Function to predict the next word based on a given word
def predict_next_word(word):
    word = word.lower()
    if word in cfd:
        next_words = cfd[word].max()
        return next_words
    else:
        return "Word not found in the corpus"
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Test the prediction function
start_word = "science"
next_word = predict_next_word(start_word)
print(f"The next word after '{start_word}' is '{next_word}'.")
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Generate a sequence of words
def generate_sequence(start_word, num_words):
    sequence = [start_word]
    currnet_word = start_word
    for _ in range(num_words):
        next_word = predict_next_word(currnet_word)
        if next_word == "Word not found in the corpus":
            break
        sequence.append(next_word)
        currnet_word = next_word
    return " ".join(sequence)
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Generate a word starting from science
generated_sequence = generate_sequence("science", 9)
print(f"Generated sequence starting from '{start_word}': {generated_sequence}")
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
words = nltk.word_tokenize(" ".join(reuters.words()))
print(len(words))
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
#Create Trigrams
tri_grams = list(trigrams(words))
print(tri_grams[:10])
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
model = defaultdict(lambda: defaultdict(lambda: 0))
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
# Count frequency of co-occurrence
for w1, w2, w3 in tri_grams:
    model[(w1, w2)][w3] += 1
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
# Transform counts into probabilities
for w1_w2 in model:
    total_count = sum(model[w1_w2].values())

    for w3 in model[w1_w2]:
        model[w1_w2][w3] /= total_count
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
# Function to predict the next word
def predict_next_word(w1, w2):
    next_words = model.get((w1, w2), {})

    if next_words:
        return max(next_words, key=next_words.get)
    else:
        return "Word not found in the corpus"
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
print(predict_next_word("the", "price"))
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
import math

def calculate_perplexity(test_sentence):
    test_tokens = word_tokenize(test_sentence)

    log_prob_sum = 0

    for i in range(2, len(test_tokens)):
        w1 = test_tokens[i - 2]
        w2 = test_tokens[i - 1]
        w3 = test_tokens[i]

        prob = model[(w1, w2)].get(w3, 1e-10)

        log_prob_sum += math.log(prob)

    perplexity = math.exp(-log_prob_sum / (len(test_tokens) - 2))

    return perplexity
```

```python
print("Performed by Mulla Huzaifa Sufyan(UIN: 231A017)")
test_sentence = "ASIAN EXPORTERS FEAR DAMAGE"

perplexity = calculate_perplexity(test_sentence)

print("Test Sentence:", test_sentence)
print("Perplexity:", perplexity)
```
